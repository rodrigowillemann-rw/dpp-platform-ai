# DPP Platform AI - Diagnostico Sintetico

Exemplo simples para calcular uma leitura inicial de prontidao para Digital Product Passport.

```python
import pandas as pd
df = pd.read_csv('../data/requisitos_dpp_textil_exemplo.csv')
peso = {'alta': 3, 'media': 2, 'baixa': 1}
maturidade = {'pendente': 0, 'parcial': 0.5, 'atendido': 1}
df['peso'] = df['criticidade'].map(peso)
df['maturidade'] = df['status_exemplo'].map(maturidade)
score = (df['peso'] * df['maturidade']).sum() / df['peso'].sum() * 100
print(f'Score sintetico de prontidao DPP: {score:.1f}%')
df[['requisito','categoria','criticidade','status_exemplo']]
```
